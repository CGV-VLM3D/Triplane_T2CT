# 06 — 번들 그룹별 복셀 소견

큐레이션된 5개 번들 그룹에 대해 미리 세팅된 셀들입니다. **all-zero**(방사선학적 정상) 소견을 이상(abnormal) 그룹들과 비교해 보세요. 폐 실질은 **lung** 윈도우, 연부조직/물질은 **mediastinal**, 골격은 **bone** 윈도우를 사용하세요. HU는 마우스 호버로 확인.

단위: 각 볼륨은 reconstruction 하나이고, 라벨은 scan 단위입니다.

## no_chest 예시

`no_chest` reconstruction(비흉부 FOV)은 **`valid_fixed`에서 제외**됩니다 — 이들은 다른 recon은 살아있는 scan에서 제거된 recon이므로, 클린 census나 번들에는 `no_chest` NIfTI가 포함되지 않습니다. 예: scan **valid_109_a**는 recon 1/2/4가 `no_chest`로 표시됨(3 & 5만 남음). 아래에서는 그런 study의 scan을 보기 위해 살아남은 흉부 recon **valid_109_a_3**을 봅니다.
