# 08 — Reconstruction 쌍 (커널 / spacing 차이)

`view_recon_pair(p1, p2)`는 *같은 scan*의 recon **_1** vs recon **_2**를 나란히 보여줍니다. orientation + window는 공유하지만 slice 슬라이더는 독립적입니다(recon_2는 흔히 더 촘촘한 spacing / 다른 커널 reconstruction이라 slice 수가 다릅니다). 두 패널에서 같은 해부학적 레벨을 맞춘 뒤 **엣지 선명도 / 노이즈 질감**을 비교하세요 — 그게 커널 차이입니다. 리포트와 라벨은 한 scan의 reconstruction들 간에 동일합니다.
